# NYC 311 Service Operations Analytics — EDA

This notebook is the exploratory-analysis workspace for the cleaned 2025 NYC 311 dataset.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA_PATH = Path('../data/processed/nyc_311_2025_clean.csv')
df = pd.read_csv(DATA_PATH, low_memory=False)
df.head()
```

```python
print(f'Rows: {len(df):,}')
print(f'Columns: {df.shape[1]}')
df.info()
```

```python
summary = pd.Series({
    'total_requests': len(df),
    'closed_rate_pct': df['is_closed'].mean() * 100,
    'avg_resolution_hours': df['resolution_hours'].mean(),
    'median_resolution_hours': df['resolution_hours'].median(),
})
summary
```

```python
top_complaints = df['complaint_type'].value_counts().head(15).sort_values()
top_complaints.plot(kind='barh', figsize=(10, 7), title='Top 15 Complaint Types')
plt.xlabel('Service Requests')
plt.tight_layout()
plt.show()
```

```python
monthly = df.groupby('created_month').size()
monthly.plot(figsize=(11, 5), marker='o', title='Monthly Service Request Trend')
plt.ylabel('Requests')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
